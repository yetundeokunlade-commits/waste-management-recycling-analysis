```python

# Select household records for co-mingled materials
household_coverage_df = household_df[
    household_df["RowText"] == "Co mingled materials"
].copy()

# Display the number of records and authorities
print("Household coverage records:", len(household_coverage_df))
print("Local authorities:", household_coverage_df["Authority"].nunique())

# Display a sample of the results
display(
    household_coverage_df[
        ["Authority", "Period", "Data_numeric"]
    ].head(10)
)
```

```python

# Calculate recorded recycling tonnage by authority and period
recycling_by_authority_period = (
    recycling_df.groupby(["Authority", "Period"])["Data_numeric"]
    .sum()
    .reset_index(name="Recycling_Tonnage")
)

# Prepare household coverage data
household_by_authority_period = (
    household_coverage_df[
        ["Authority", "Period", "Data_numeric"]
    ]
    .rename(columns={"Data_numeric": "Households_Covered"})
)

# Match recycling tonnage with household coverage
coverage_comparison_df = pd.merge(
    recycling_by_authority_period,
    household_by_authority_period,
    on=["Authority", "Period"],
    how="inner"
)

# Display the number of matched records and authorities
print("Matched authority-period records:", len(coverage_comparison_df))
print("Local authorities included:", coverage_comparison_df["Authority"].nunique())

# Display a sample of the comparison
display(coverage_comparison_df.head(10))
```

```python

# Calculate the correlation between household coverage and recycling tonnage
coverage_correlation = coverage_comparison_df[
    ["Households_Covered", "Recycling_Tonnage"]
].corr().loc["Households_Covered", "Recycling_Tonnage"]

# Display the correlation result
print(
    "Correlation between household coverage and recorded recycling tonnage:",
    round(coverage_correlation, 3)
)
```

### 6.40 Findings from Recycling Tonnage and Household Coverage

I compared recorded recycling tonnage with the number of households reported as receiving a co-mingled materials collection across 282 local authorities, using 1,124 matched authority-period records.

The Pearson correlation coefficient was **0.757**, indicating a strong positive relationship between the two measures. This suggests that authorities reporting higher household coverage generally also recorded higher recycling tonnage.

However, correlation does not prove causation. Larger authorities may have more households and generate more recycling tonnage. The comparison also uses household coverage for co-mingled materials, rather than total household coverage across all collection types. Therefore, the result should be interpreted as an association between these two recorded measures, not as proof that increased household coverage causes higher recycling tonnage.
